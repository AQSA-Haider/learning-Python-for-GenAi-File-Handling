# Writing Sales record to a file
# Firstly storing the the sales as list

```python
sales=[1200,450,980,1500,3000]
```

# Opening a file named sales_data.txt in write mode('w') and writing each sale on a new line,
# and then closing the file. 

```python
f=open('sales_data.txt','w')
f.write('sales=\n' '1200\n' '450\n' '980\n' '1500\n' '3000')
f.close()
```

# Reopening the file to print its content :

```python
f=open('sales_data.txt','r')
content=f.read()
print(content)
f.close()
```

# Optional: writing the data in comma separated format instead of separate lines:

```python
f=open('sales_data.txt','w')
f.write('sales= 1200, 450, 980, 1500, 3000')
f.close()
```

# Opening the file to print the updated content:

```python
f=open('sales_data.txt','r')
content=f.read()
print(content)
f.close()
```
