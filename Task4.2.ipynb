# Read File in different format using the same file sales_data.txt
# Rewriting the data into the file 

```python
with open('sales_data.txt','w')as  f:
    line=['1200\n', '450\n', '980\n', '1500\n', '3000\n']
    f.writelines(line)
    f.close()
```

# Reading the entire file using .read() and printing it:

```python
f=open('sales_data.txt','r')
content=f.read()
print(content)
f.close()
```

# Reading the first line using .readline()

```python
f=open('sales_data.txt','r')
print(f.readline())
```

# Reading all the lines using .readlines() and converting them into a list of integers

```python
with open ('sales_data.txt','r')as f:
    line=f.readlines() 
    new=[int(x) for x in line]
    print(new)
    f.close()
```
