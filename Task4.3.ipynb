# Appending new sales to a file

```python
f=open('sales_data.txt','a')
new=['5000\n', '2500\n', '1700\n' ]
f.writelines(new)
f.close()
```

# Reopening and printing the entire updated file

```python
f=open('sales_data.txt','r')
content=f.read()
print(content)
f.close()
```

# Optional: Printing the total no. of lines in the file :

```python
f=open('sales_data.txt','r')
content = f.readlines()
print(len(content))
f.close()
```
