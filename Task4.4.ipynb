# Converting the data into integers:

```python
f=open('sales_data.txt','r')
read=f.readlines()
integer=[int(x) for x in read]
print(integer)
```

# Calculating and Printing:
# ~Total Sales
# ~Highest Sales
# ~Lowest Sales
# ~Average Sales 
without use of advanced libraries

```python
sales=0
for i in integer:
    sales=sales+i
    avg=sales/len(integer)
    highest=max(integer)
    lowest=min(integer)
print("Total Sales=",sales)
print("Highest Sales=",highest)
print("Lowest Sales=",lowest)
print("Average Sales=",avg)
```
