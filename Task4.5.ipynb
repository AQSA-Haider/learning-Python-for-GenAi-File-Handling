# Task-4.5: Create product info file(user input)
# Asking the user for 3 product names and thier prices
# Writing them into a file named product.txt

```python

with open("products.txt",'w') as file:
        file.write("Product | Price\n")
        for i in range(3):
                k=input(f"enter product name{i+1}:")
                v=input(f"enter product price{i+1}:")
                file.write(f"{k} | {v}\n")
        
print("\n----Product List----")
```

## Reading the content from file product.txt

```python
with open("products.txt",'r')as f:
    content=f.read()
    print(content)
```
