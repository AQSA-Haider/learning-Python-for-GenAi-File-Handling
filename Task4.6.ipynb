# Read file safely
# Asking the user for a filname to open
# if the file exists it reads and prints it .
# if not it prints a message

```python
ask=input("Enter the file name ?")
if os.path.exists(ask)==True:
    f=open(f'{ask}','r')
    content=f.read()
    print(content)
else:
    print("File not found,Please check filename")
```
