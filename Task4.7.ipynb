# firstly, Creating a dictionary

```python
prices={"Mouse":500,"Keyboard":800,"Monitor":7000,"Pendrive":400,"Camera":5000}
```

# then, asking the user for a discounted price
# and writing the discounted price into discount_report.txt
# in the format:
# Product | Original Price | Discounted Price

```python
discount=int(input("Enter discount percentage :"))
length=0
for i in prices.values():
    length=length+1
print(length)
with open("discount_report.txt","w") as file:
    file.write("Product | Original Price | Discounted Price\n")
    for i,j in prices.items():
        new_price= j-(j*discount/100)
        file.write(f"{i} | {j} | {new_price}\n")
```
